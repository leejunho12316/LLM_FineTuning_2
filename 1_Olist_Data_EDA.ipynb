##2. orders + order_payments & order_reviews & order_items

- ord & ord_pay : 하나의 주문을 여러 결제수단으로 나누어 결제하는 경우 order_id의 중복이 생김.

```python
pd.merge(left = df_ord, right = df_ord_pay, on='order_id', how='left')
```

- ord & ord_rev : order_id 별 review 2개 이상인것 꽤 있음. 하지만 65%는 review 값이 NaN. 아마 오류로 NaN 들어간 다음에 정상 review가 들어간 것 같음.

```python
temp = pd.merge(left = df_ord, right = df_ord_rev, on='order_id', how='left')
temp_dup = temp[temp['order_id'].duplicated(keep=False)].sort_values(by='order_id')
sum(temp_dup['review_comment_message'].isnull()) / len(temp_dup)
```
